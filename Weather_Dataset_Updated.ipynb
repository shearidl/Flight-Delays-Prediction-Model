# Cleaning U.S. weather dataset & EDA

```python
df.info()
```

```python
# dropping irrelevant columns from df

df.drop(columns=["TEMP_ATTRIBUTES", "DEWP_ATTRIBUTES", "SLP_ATTRIBUTES", "STP_ATTRIBUTES", 
                 "VISIB_ATTRIBUTES", "WDSP_ATTRIBUTES", "MAX_ATTRIBUTES", "MIN_ATTRIBUTES",
                 "PRCP_ATTRIBUTES", "COUNTRY"], axis=1, inplace=True)
```

```python
# replacing missing values with NaN (according to documentation)

df = df.replace(9999.9, np.nan)
df = df.replace(999.9, np.nan)
df = df.replace(99.9, np.nan)
```

```python
df.info()
```

```python
# creating weather type columns

df["FRSHTT"] = df["FRSHTT"].astype(str).str.zfill(6)

df[["Fog", "Rain", "Snow", "Hail", "Thunder", "Tornado"]] = (
    df["FRSHTT"]
    .apply(list)          
    .apply(pd.Series) 
    .astype(int)
)
```

```python
df
```

# Combine with Cities Dataset

```python
# locations of cities by latitude and longitude

# https://simplemaps.com/data/us-cities

cities = pd.read_csv('uscities.csv')

cols = ["city", "state_id", "state_name", "lat", "lng"]
cities = cities[cols]
cities
```

```python
import geopandas as gpd
from shapely.geometry import Point

# using geopandas to turn longitudes and latitudes into a single point

gdf_weather = gpd.GeoDataFrame(
    df, geometry=gpd.points_from_xy(df.LONGITUDE, df.LATITUDE), crs="EPSG:4326"
)

gdf_cities = gpd.GeoDataFrame(
    cities, geometry=gpd.points_from_xy(cities.lng, cities.lat), crs="EPSG:4326"
) 
    
```

```python
# using geopandas to combine cities dataset with longitudes and latitudes from df/weather dataset
# need to specify variation in longitudes and latitudes to combine
weather_geo = gdf_weather.to_crs(epsg=32631)
cities_geo = gdf_cities.to_crs(epsg=32631)

joined_weather_cities = gpd.sjoin_nearest(weather_geo, cities_geo, how="inner", max_distance=35000)
# choosing 35000m ~ 21 miles distance from point

joined_weather_cities = joined_weather_cities[joined_weather_cities["state_name"] != "Puerto Rico"]
# removing Puerto Rico to focus only on 50 states

joined_weather_cities
```

```python
joined_weather_cities = joined_weather_cities.reset_index(drop=True)
joined_weather_cities
```
